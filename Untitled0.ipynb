```python
from src.chunking import (
    HeadingSectionChunker,
    SentenceChunker,
    RecursiveChunker,
)
from src.models import Document
from src.store import EmbeddingStore

BASE = Path("data/university")
TOP_K = 3

BENCHMARKS = [
    {
        "id": 1,
        "query": "Quy trình hoãn thi giữa kỳ cần thực hiện như thế nào?",
        "gold_doc": "quy-che-dao-tao-chinh-quy",
        "gold_answer": (
            "nộp đơn kèm minh chứng cho Phòng Đào tạo Đại học "
            "trong vòng 03 ngày kể từ ngày thi"
        ),
        "metadata_filter": {"audience": "student"},
    },
    {
        "id": 2,
        "query": (
            "Bộ phận nào của Trường sẽ xem xét các trường hợp "
            "có lí do chính đáng để vắng thi giữa kỳ?"
        ),
        "gold_doc": "quy-dinh-to-chuc-thi",
        "gold_answer": "P.ĐTĐH",
        "metadata_filter": {},
    },
    {
        "id": 3,
        "query": (
            "Thời hạn lưu trữ đề thi các môn học hệ đại học chính quy "
            "của Trường là bao lâu?"
        ),
        "gold_doc": "quy-dinh-khoa-luan-tot-nghiep",
        "gold_answer": "9 năm",
        "metadata_filter": {},
    },
    {
        "id": 4,
        "query": "Sinh viên thuộc chương trình tài năng có các hình thức nào?",
        "gold_doc": "quy-trinh-phan-cong-can-bo-coi-thi",
        "gold_answer": "chính thức và dự bị",
        "metadata_filter": {},
    },
    {
        "id": 5,
        "query": "KLTN là viết tắt của cụm từ nào?",
        "gold_doc": "quy-dinh-khoa-luan-tot-nghiep",
        "gold_answer": "Khóa luận tốt nghiệp",
        "metadata_filter": {},
    },
]
```

```python
def parse_markdown(path):
    text = path.read_text(encoding="utf-8")

    if "---" in text:
        _, frontmatter, body = text.split("---", 2)
    else:
        frontmatter = ""
        body = text

    metadata = {}

    for line in frontmatter.strip().splitlines():
        if ":" in line:
            key, value = line.split(":", 1)
            metadata[key.strip()] = value.strip().strip('"')

    return metadata, body.strip()


def build_store(strategy_name, chunker, embedder):
    store = EmbeddingStore(
        collection_name=f"university_{strategy_name}",
        embedding_fn=embedder,
    )

    total_chunks = 0
    document_count = 0

    for path in sorted(BASE.glob("*.md")):
        if path.name == "README.md":
            continue

        metadata, body = parse_markdown(path)
        chunks = chunker.chunk(body)

        document_count += 1
        total_chunks += len(chunks)

        documents = []

        for index, chunk in enumerate(chunks):
            documents.append(
                Document(
                    id=f"{metadata['doc_id']}::chunk-{index}",
                    content=chunk,
                    metadata={
                        **metadata,
                        "doc_id": metadata["doc_id"],
                        "chunk_index": index,
                    },
                )
            )

        store.add_documents(documents)

    print(
        f"{strategy_name}: "
        f"{document_count} documents, "
        f"{total_chunks} chunks"
    )

    return store
```

```python
stores = {}

strategies = {
    "heading_section": HeadingSectionChunker(
        chunk_size=500,
    ),
    "sentence": SentenceChunker(
        max_sentences_per_chunk=3,
    ),
    "recursive": RecursiveChunker(
        chunk_size=500,
    ),
}

for strategy_name, chunker in strategies.items():
    stores[strategy_name] = build_store(
        strategy_name,
        chunker,
        embedder,
    )
```

```python
def normalize(text):
    import re
    text = text.lower().strip()
    text = re.sub(r"\s+", " ", text)
    text = re.sub(r"[.,;:!?]", "", text)
    return text


def evaluate_agent_answer(agent_answer, gold_answer):
    if not agent_answer:
        return False
    return normalize(gold_answer) in normalize(agent_answer)


def run_benchmark(strategy_name, store, llm_fn=None):
    print("\n" + "=" * 80)
    print(f"STRATEGY: {strategy_name}")
    print("=" * 80)

    retrieval_hits = 0
    answer_hits = None if llm_fn is None else 0

    for item in BENCHMARKS:
        results = store.search_with_filter(
            query=item["query"],
            top_k=TOP_K,
            metadata_filter=item["metadata_filter"],
        )

        hit_rank = None
        for rank, result in enumerate(results, start=1):
            same_document = result["metadata"].get("doc_id") == item["gold_doc"]
            contains_gold = item["gold_answer"].lower() in result["content"].lower()
            if same_document and contains_gold:
                hit_rank = rank
                break

        retrieval_hit = hit_rank is not None
        retrieval_hits += int(retrieval_hit)

        agent_answer = None
        answer_hit = None
        if llm_fn is not None:
            context = "\n\n".join(
                f"[Source {i}]\n{r['content']}"
                for i, r in enumerate(results, start=1)
            )
            prompt = (
                "Bạn là trợ lý trả lời chỉ dựa trên ngữ cảnh. "
                "Nếu không đủ thông tin, hãy nói không tìm thấy.\n\n"
                f"Ngữ cảnh:\n{context}\n\n"
                f"Câu hỏi: {item['query']}\nCâu trả lời:"
            )
            agent_answer = llm_fn(prompt)
            answer_hit = evaluate_agent_answer(agent_answer, item["gold_answer"])
            if answer_hits is not None:
                answer_hits += int(answer_hit)

        print(f"\nQuery {item['id']}: retrieval={'HIT' if retrieval_hit else 'MISS'}")
        print("Filter:", item["metadata_filter"])
        print("Gold document:", item["gold_doc"])
        print("Gold rank:", hit_rank)
        if llm_fn is None:
            print("Agent answer: chưa cấu hình llm_fn")
        else:
            print("Agent answer:", "CORRECT" if answer_hit else "INCORRECT")
            print("Answer:", agent_answer)
        for rank, result in enumerate(results, start=1):
            print(
                f"  {rank}. {result['metadata'].get('doc_id')} "
                f"(chunk={result['metadata'].get('chunk_index')}, "
                f"score={result['score']:.4f})"
            )

    print(f"\nRetrieval total: {retrieval_hits}/5")
    if llm_fn is not None:
        print(f"Agent answer total: {answer_hits}/5")
    return retrieval_hits, answer_hits
```

```python
all_results = {}

for strategy_name, store in stores.items():
    all_results[strategy_name] = run_benchmark(
        strategy_name,
        store,
        llm_fn=None,  # truyền hàm LLM thật vào đây nếu có
    )
```

```python
FILTER_QUERY = (
    "Quy trình hoãn thi giữa kỳ cần thực hiện như thế nào?"
)

for strategy_name, store in stores.items():
    print("\n" + "=" * 80)
    print(f"METADATA FILTER TEST: {strategy_name}")
    print("=" * 80)

    for audience in ["student", "faculty"]:
        results = store.search_with_filter(
            query=FILTER_QUERY,
            top_k=3,
            metadata_filter={"audience": audience},
        )

        print(f"\nAudience: {audience}")

        for rank, result in enumerate(results, start=1):
            print(
                f"{rank}. "
                f"{result['metadata'].get('doc_id')} "
                f"(score={result['score']:.4f})"
            )
            print(result["content"][:300])
            print("---")
```

```python
print("\n" + "=" * 80)
print("SUMMARY")
print("=" * 80)

for strategy_name, result in all_results.items():
    retrieval_hits, answer_hits = result

    print(
        f"{strategy_name}: "
        f"retrieval={retrieval_hits}/{len(BENCHMARKS)} "
        f"({retrieval_hits / len(BENCHMARKS) * 100:.1f}%)"
    )

    if answer_hits is None:
        print(f"{strategy_name}: agent_answer=N/A")
    else:
        print(
            f"{strategy_name}: "
            f"agent_answer={answer_hits}/{len(BENCHMARKS)}"
        )
```
